# Sentiment Analysis - Amazon Product Reviews

---
## Import Modules

```python
import pandas as pd
import spacy
from spacytextblob.spacytextblob import SpacyTextBlob
import re
```

---
## Load And Preprocess The Data

```python
# Load the dataset
df = pd.read_csv('Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products_May19.csv')
# Drop rows with missing reviews
df = df.dropna(subset=['reviews.text'])
# Select the 'reviews.text' column and rename it to 'review'
df = df[['reviews.text']].copy()
df.rename(columns={'reviews.text': 'review'}, inplace=True)
df.head()
```

---
## Load the spaCy model and add TextBlob

```python
nlp = spacy.load('en_core_web_md')
nlp.add_pipe('spacytextblob')
```

---
## Text Cleaning

```python
# Text cleaning function
def clean_text(text):
    text = str(text).lower().strip()
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)  # remove punctuation
    return text

# Remove stop words
def preprocess(doc):
    return " ".join([token.text for token in doc if not token.is_stop])
```

---
## Create A Function For Sentiment Analysis

```python
# Sentiment analysis function
def analyze_sentiment(text):
    doc = nlp(str(text).lower().strip())
    polarity = doc._.blob.polarity
    if polarity > 0:
        sentiment = 'Positive'
    elif polarity < 0:
        sentiment = 'Negative'
    else:
        sentiment = 'Neutral'
    return sentiment, polarity
```

---
## Test The Model On Sample Product Reviews

```python
# Test the function on 5 sample reviews
sample_reviews = df['review'].sample(5, random_state=42)
for review in sample_reviews:
    sentiment, polarity = analyze_sentiment(review)
    print(f"Review: {review[:100]}...")
    print(f"Sentiment: {sentiment} (Polarity: {polarity:.2f})\n")

# Compare similarity between two reviews
review1 = df['review'].iloc[0]
review2 = df['review'].iloc[1]

doc1 = nlp(clean_text(review1))
doc2 = nlp(clean_text(review2))
similarity = doc1.similarity(doc2)
print(f"Similarity between review 0 and 1: {similarity:.2f}")    
```
